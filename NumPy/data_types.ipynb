Data Types in NumPy
NumPy has some extra data types, and refer to data types with one character, like i for integers, u for unsigned integers etc.

Below is a list of all data types in NumPy and the characters used to represent them.

i - integer
b - boolean
u - unsigned integer
f - float
c - complex float
m - timedelta
M - datetime
O - object
S - string
U - unicode string
V - fixed chunk of memory for other type ( void )

```python
import numpy as np

arr = np.array([1, 2, 3, 4])

print(arr.dtype)
```

```python
import numpy as np

arr = np.array(['apple', 'banana', 'cherry'])

print(arr.dtype)
```

We use the array() function to create arrays, this function can take an optional argument: dtype that allows us to define the expected data type of the array elements:

```python
arr = np.array([1, 2, 3, 4], dtype='S')

print(arr)
print(arr.dtype)
```

```python
# The statement dtype='S' tells NumPy to store the array elements as fixed-length
# byte strings instead of integers or Unicode strings. When the integer values
# [1, 2, 3, 4] are passed, NumPy automatically converts them into byte strings,
# resulting in the array [b'1', b'2', b'3', b'4']. The prefix 'b' indicates that
# the values are stored as bytes. When we print arr.dtype, the output is '|S1',
# where 'S' represents a byte string data type, '1' indicates that each element
# occupies a maximum of 1 byte (since each number is a single character), and '|'
# means byte order is not applicable. The 'S' data type is mainly used for storing
# binary or ASCII string data, while the 'U' data type is preferred for storing
# Unicode strings in modern Python.
```

```python
arr = np.array([1, 2, 3, 4], dtype='i4')

print(arr)
print(arr.dtype)
```

The best way to change the data type of an existing array, is to make a copy of the array with the astype() method.

The astype() function creates a copy of the array, and allows you to specify the data type as a parameter.

The data type can be specified using a string, like 'f' for float, 'i' for integer etc. or you can use the data type directly like float for float and int for integer.
